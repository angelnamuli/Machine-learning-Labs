```python
np.random.seed(42)
n = 30
```

```python
#size per sqm
size_sqm = np.random.randint(30, 400, n)
size_sqm[:5]
```

```python
#no of bedrooms
bedrooms = np.random.randint(1,7,n)
bedrooms[:6]
```

```python
#no of bathrooms
bathrooms = np.random.randint(1,4,n)
bathrooms[:3]
```

```python
#year built
year_built = np.random.randint(1990,2025,n)
year_built[:6]
```

```python
#property type
property_types = ['Apartment','Bungalow','Mansion']
property_type = np.random.choice(property_types,n)
property_type[:3]
```

```python
#location
locations = ['Surburb','Village','Town','Hilly area']
location = np.random.choice(locations, n)
location[:6]
```

```python
#price meaning larger houses are more expensive
price_in_UGX_per_sqm = np.random.randint(650_000,1_000_000,n)
price = size_sqm * price_in_UGX_per_sqm
price[:3]
```

```python
#creating the dataframe
exercise_df = pd.DataFrame({
    'Price_UGX':price,
    'Size_sqm':size_sqm,
    'bedrooms':bedrooms,
    'bathrooms':bathrooms,
    'year-built':year_built,
    'location':location,
    'property_type':property_type
    
})
```

```python
exercise_df.shape
```

```python
exercise_df.dtypes
```

```python
exercise_df.head(8)
```

```python
#part b
filtered_df = exercise_df[
    (exercise_df['location'].isin(['Village','Town'])) &
    (exercise_df['Price_UGX'] > 50_000_000) &
    (exercise_df['bedrooms'] >= 3)
]
```

```python
#reset index
filtered_df = filtered_df.reset_index(drop=True)
filtered_df

```

```python
#average price of the filtered
filtered_df['Price_UGX'].mean()
```

```python
#most frequent property type
filtered_df['property_type'].mode()[0]
```

```python
# rows 5 to 14(inclusive)
extracted = exercise_df.iloc[5:15,2:4]
extracted
```

```python
extracted_df = extracted.rename(columns={
    'bedrooms':'num-bedrooms',
    'bathrooms':'num-bathrooms'
})
extracted_df
```

```python
#mean 
extracted_df.mean()
```

```python
#standard deviation
extracted_df.std()
```

```python
#row with highest value
extracted_df.loc[extracted_df['num-bedrooms'].idxmax()]
```

```python
#approach 1
exercise_df['first-deposit']= exercise_df['Price_UGX'] * 0.10
exercise_df.head()
```

```python
#approach 2
exercise_df['second-deposit']= exercise_df['Price_UGX'] /10
exercise_df.head()
```

```python
#approach 3
exercise_df['third-deposit']= exercise_df['Price_UGX'].apply(lambda x: x * 0.10)
exercise_df.head()
```

```python
#approach 4
exercise_df = exercise_df.assign(
    deposit_4 = exercise_df['Price_UGX'] * 0.10
) 
exercise_df.head()
```

```python
#comparison of the approaches
(
    (exercise_df["first-deposit"] == exercise_df["second-deposit"]) &
    (exercise_df["first-deposit"] == exercise_df["third-deposit"]) &
    (exercise_df["first-deposit"] == exercise_df["deposit_4"])
).all()
```
